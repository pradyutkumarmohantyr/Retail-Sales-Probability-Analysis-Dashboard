```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/sample_-_superstore.csv", encoding="latin1")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace(" ", "_")
    .str.replace("-", "_")
    .str.lower()
)
df.columns.tolist()
```

```python
df["order_date"] = pd.to_datetime(df["order_date"], format="%d-%m-%Y")
df["ship_date"] = pd.to_datetime(df["ship_date"], format="%d-%m-%Y")

df[["order_date", "ship_date"]].head()
```

```python
# Sales should never be negative
print("Negative sales:", (df["sales"] < 0).sum())

# Ship date should never be before order date
print("Ship before order:", (df["ship_date"] < df["order_date"]).sum())
```

```python
df["shipping_days"] = (df["ship_date"] - df["order_date"]).dt.days
df["order_year"] = df["order_date"].dt.year
df["order_month"] = df["order_date"].dt.month
df["profit_margin"] = df["profit"] / df["sales"]

df[["shipping_days", "order_year", "order_month", "profit_margin"]].head()
```

```python
df.to_csv("../data/processed/superstore_clean.csv", index=False)
print("Saved:", df.shape)
```

```python
print("Invalid discount:", ((df["discount"] < 0) | (df["discount"] > 1)).sum())
print("Non-positive quantity:", (df["quantity"] <= 0).sum())
```
